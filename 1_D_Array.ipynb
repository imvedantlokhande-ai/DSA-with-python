```python
class One_D_Array:
    def __init__(self, initial_size=10):
        self.size = initial_size
        self.array = [None] * self.size
        self.length = 0  # Current number of elements

    def display(self):

        if self.length == 0:
            print("Array is empty")
            return
        print("Current Array:", end=" ")
        for i in range(self.length):
            print(self.array[i], end=" ")
        print()
        print(f"Size: {self.size}, Length: {self.length}")

    def insert_at_end(self, value):
        if self.length >= self.size:
            print("Array is full! Cannot insert")
            return False

        self.array[self.length] = value
        self.length += 1
        print(f"Inserted {value} at end")
        return True

    def insert_at_position(self, position, value):

        if position < 0 or position > self.length:
            print(f"Invalid position! Must be between 0 and {self.length}")
            return False

        if self.length >= self.size:
            print("Array is full! Cannot insert")
            return False

        # Shift elements to the right
        for i in range(self.length, position, -1):
            self.array[i] = self.array[i - 1]

        self.array[position] = value
        self.length += 1
        print(f"Inserted {value} at position {position}")
        return True

    def delete_at_end(self):

        if self.length == 0:
            print("Array is empty! Cannot delete")
            return None

        deleted_value = self.array[self.length - 1]
        self.array[self.length - 1] = None
        self.length -= 1
        print(f"Deleted {deleted_value} from end")
        return deleted_value

    def delete_at_position(self, position):

        if self.length == 0:
            print("Array is empty! Cannot delete")
            return None

        if position < 0 or position >= self.length:
            print(f"Invalid position! Must be between 0 and {self.length - 1}")
            return None

        deleted_value = self.array[position]


        for i in range(position, self.length - 1):
            self.array[i] = self.array[i + 1]

        self.array[self.length - 1] = None
        self.length -= 1
        print(f"Deleted {deleted_value} from position {position}")
        return deleted_value

    def search(self, value):

        for i in range(self.length):
            if self.array[i] == value:
                print(f"Found {value} at position {i}")
                return i
        print(f"Value {value} not found in array")
        return -1

    def search_all_occurrences(self, value):

        positions = []
        for i in range(self.length):
            if self.array[i] == value:
                positions.append(i)

        if positions:
            print(f"Found {value} at positions: {positions}")
            return positions
        else:
            print(f"Value {value} not found in array")
            return []
def array_menu_1d():

    arr = One_D_Array(5)
    print("\n" + "=" * 40)
    print("1D ARRAY OPERATIONS MENU")
    print("=" * 40)

    while True:
        print("\n1. Insert at end")
        print("2. Insert at position")
        print("3. Delete at end")
        print("4. Delete at position")
        print("5. Search value")
        print("6. Search all occurrences")
        print("7. Display array")
        print("8. Exit")

        choice = input("\nEnter your choice (1-8): ")

        if choice == '1':
            value = input("Enter value to insert: ")
            arr.insert_at_end(value)

        elif choice == '2':
            pos = int(input("Enter position: "))
            value = input("Enter value to insert: ")
            arr.insert_at_position(pos, value)

        elif choice == '3':
            arr.delete_at_end()

        elif choice == '4':
            pos = int(input("Enter position: "))
            arr.delete_at_position(pos)

        elif choice == '5':
            value = input("Enter value to search: ")
            arr.search(value)

        elif choice == '6':
            value = input("Enter value to search: ")
            arr.search_all_occurrences(value)

        elif choice == '7':
            arr.display()

        elif choice == '8':
            print("Thank you for using 1D Array Operations!")
            break

        else:
            print("Invalid choice! Please try again.")
if __name__ == "__main__":
    print("=" * 20)
    print("ARRAY OPERATIONS")
    print("=" * 20)
    array_menu_1d()
```
